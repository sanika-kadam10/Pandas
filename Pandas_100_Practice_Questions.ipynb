### Q4. Perform arithmetic operations on a Series.
> 💡 **Hint:** Use `+`, `-`, `*`, `/`, `//`, `%`, `**` directly on the Series. Operations are **element-wise** and **index-aligned**.

---

```python
s = pd.Series([10, 20, 30, 40])

print("Addition:")
print(s + 5)

print("\nSubtraction:")
print(s - 5)

print("\nMultiplication:")
print(s * 2)

print("\nDivision:")
print(s / 2)

print("\nFloor Division:")
print(s // 3)

print("\nModulus:")
print(s % 3)

print("\nPower:")
print(s ** 2)
```

### Q56. Detect outliers in a column.
> 💡 **Hint:** Use the **IQR method**: compute Q1, Q3, IQR = Q3–Q1. Outliers are values below `Q1 - 1.5*IQR` or above `Q3 + 1.5*IQR`. Alternatively use Z-score: `|z| > 3`.

---

```python
import pandas as pd
import numpy as np
df = pd.DataFrame({
    'salary': [30000, 35000, 40000, 45000, 50000, 200000]
})
q1=df["salary"].quantile(0.25)
q3=df["salary"].quantile(0.75)
IQR=q3-q1
lower=q1-1.5*IQR
upper=q3+1.5*IQR
outliers = df[(df['salary'] < lower) | (df['salary'] > upper)]

print(outliers)
```

### Q61. Filter rows using AND condition.
> 💡 **Hint:** `df[(df["col1"] > val) & (df["col2"] == "x")]` — wrap each condition in parentheses and use `&`.

---

```python
df = pd.DataFrame({
    "Age": [20, 25, 30, 35],
    "City": ["Pune", "Mumbai", "Pune", "Delhi"]
})
result=df[(df["Age"]>25) & (df["City"]=="Pune")]
print(result)
```

### Q62. Filter rows using OR condition.
> 💡 **Hint:** `df[(df["col1"] > val) | (df["col2"] == "x")]` — use `|` between conditions.

---

```python
df = pd.DataFrame({
    "Age": [20, 25, 30, 35],
    "City": ["Pune", "Mumbai", "Pune", "Delhi"]
})
result=df[(df["Age"]>30) | (df["City"]=="Pune")]
print(result)
```

### Q63. Filter rows using NOT condition.
> 💡 **Hint:** `df[~(df["col"] == "val")]` — use `~` to negate a boolean condition.

---

```python
df = pd.DataFrame({
    "Name": ["Sanika", "Rahul", "Amit", "Priya"],
    "City": ["Pune", "Mumbai", "Pune", "Delhi"]
})

result = df[~(df["City"] == "Pune")]

print(result)
```
